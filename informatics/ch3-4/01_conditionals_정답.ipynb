# 조건문


```python
speed = int(input("속도: "))
if speed > 30:
    print("과속입니다.")
```

```python
height = float(input("키: "))
if height < 140:
    print("더블스핀 놀이기구 탑승 불가")
```

```python
road = input("도로 종류(어린이 보호 구역이면 a): ")
if road == 'a':
    print("어린이 보호 구역은 시속 30km입니다.")
else:
    print("일반 도로는 시속 50km입니다.")
```

```python
age = int(input("나이를 입력하세요."))
if age <= 18:
    print("입장권 가격은 40,000원입니다.")
else:
    print("입장권 가격은 50,000원입니다.")
```

```python
score = int(input("점수: "))
if score >= 80:
    print("A")
elif score >= 60:
    print("B")
else:
    print("C")
```

```python
height = int(input('키: '))
if height < 100:
    print('탑승 불가')
elif height < 130:
    print('보호자 동행 시 탑승 가능')
elif height < 195:
    print('탑승 가능')
else:
    print('탑승 불가')
```
