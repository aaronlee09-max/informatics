# 지진 데이터 분석


```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv('earthquake.csv', encoding='utf-8')
print(df.head())
print(df.isnull().sum())
df.dropna(axis=0, inplace=True)
df['year'] = pd.to_datetime(df['시간']).dt.year
print(df['year'].value_counts().sort_index())
sns.scatterplot(data=df, x='year', y='규모')
df['지역'] = df['위치'].str[:2]
counts = df['지역'].value_counts()
plt.figure(figsize=(8, 8))
plt.pie(counts, labels=counts.index, autopct='%.1f%%')
plt.show()
```
