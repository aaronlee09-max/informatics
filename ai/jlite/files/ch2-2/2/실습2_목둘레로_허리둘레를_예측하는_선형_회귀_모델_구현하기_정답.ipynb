```python
# 실습에 필요한 라이브러리 준비 (Colab + JupyterLite 호환)
import sys

_packages = [
    ("pandas", "pandas"),
    ("numpy", "numpy"),
    ("matplotlib", "matplotlib"),
    ("seaborn", "seaborn"),
    ("sklearn", "scikit-learn"),
]

if sys.platform == "emscripten":
    import piplite
    await piplite.install([pip_name for _, pip_name in _packages])
else:
    import importlib, subprocess
    for module_name, pip_name in _packages:
        try:
            importlib.import_module(module_name)
        except ModuleNotFoundError:
            subprocess.run([sys.executable, "-m", "pip", "install", "-q", pip_name], check=False)

from sklearn import set_config
set_config(display="text")

print("라이브러리 준비 완료")
```

```python
import pandas as pd
df = pd.read_csv('jeans.csv')
df.head()
```

```python
X = df[['neck']].values
y = df['waist'].values
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
print(X_train.shape, X_test.shape, y_train.shape, y_test.shape)
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(X_train, y_train)
print(model.coef_, model.intercept_)
```

```python
print(model.score(X_train, y_train))
print(model.score(X_test, y_test))
```

```python
predictions = model.predict(X_test)
print(predictions[:5])
print(y_test[:5])
```

```python
import matplotlib.pyplot as plt
plt.scatter(X_test, y_test, color='blue', label='Actual')
plt.plot(X_test, predictions, color='red', label='Predicted')
plt.xlabel('Neck')
plt.ylabel('Waist')
plt.legend()
plt.show()
```
